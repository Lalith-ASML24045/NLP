# Document to Vector Conversion (NLP)

This notebook demonstrates 3 common ways to convert text documents into numeric vectors:

1. **TF-IDF** (scikit-learn) — most common, fast, interpretable
2. **Bag-of-Words** (word counts) — simplest baseline
3. **Sentence Embeddings** (sentence-transformers) — captures semantic meaning

## 1. TF-IDF Vectors

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np

# Sample documents
documents = [
    "Natural language processing is fun",
    "I love learning about machine learning",
    "Text vectorization converts words into numbers",
    "Deep learning powers modern NLP models"
]

# Create the vectorizer
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words='english',
    ngram_range=(1, 1)  # use (1,2) for unigrams + bigrams
)

# Fit and transform documents into vectors
tfidf_matrix = vectorizer.fit_transform(documents)

# Convert to a dense array to inspect
vectors = tfidf_matrix.toarray()
feature_names = vectorizer.get_feature_names_out()

print("Vocabulary:", feature_names)
print("\nDocument vectors shape:", vectors.shape)

for i, doc in enumerate(documents):
    print(f"\nDoc {i}: {doc}")
    print("Vector:", np.round(vectors[i], 3))
```

## 2. Bag-of-Words (raw word counts)

```python
from sklearn.feature_extraction.text import CountVectorizer

bow_documents = [
    "Natural language processing is fun",
    "I love learning about machine learning"
]

count_vectorizer = CountVectorizer()
count_matrix = count_vectorizer.fit_transform(bow_documents)

print("Vocabulary:", count_vectorizer.get_feature_names_out())
print("Vectors:\n", count_matrix.toarray())
```

## 3. Semantic Embeddings (sentence-transformers)

Better for similarity search, clustering by meaning, and semantic search tasks.

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer('all-MiniLM-L6-v2')

embed_documents = [
    "Natural language processing is fun",
    "I love learning about machine learning",
    "The weather is nice today"
]

embeddings = model.encode(embed_documents)
print("Embedding shape:", embeddings.shape)  # e.g. (3, 384)

# Cosine similarity between doc 0 and doc 1
sim = cosine_similarity([embeddings[0]], [embeddings[1]])
print("Similarity (doc0 vs doc1):", sim[0][0])

sim2 = cosine_similarity([embeddings[0]], [embeddings[2]])
print("Similarity (doc0 vs doc2):", sim2[0][0])
```

## Which one should you use?

| Method | Best for | Notes |
|---|---|---|
| TF-IDF | Classic ML (classification, clustering) | Fast, interpretable |
| Bag-of-Words | Simple baselines | Raw counts, less nuanced |
| Sentence Embeddings | Semantic search, similarity | Captures meaning, needs a model download |
