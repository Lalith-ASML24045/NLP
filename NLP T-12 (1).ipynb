```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering

# Movie data
movies = [
    "Movie A", "Movie B", "Movie C", "Movie D",
    "Movie E", "Movie F", "Movie G", "Movie H"
]

# Features: Action, Comedy, Romance, Thriller
data = np.array([
    [9, 2, 1, 8],
    [8, 3, 2, 9],
    [2, 9, 8, 2],
    [3, 8, 9, 1],
    [7, 2, 3, 8],
    [2, 8, 7, 2],
    [8, 3, 2, 7],
    [3, 9, 8, 2]
])

# -------------------------------
# Agglomerative Clustering
# -------------------------------

model = AgglomerativeClustering(n_clusters=2, linkage='ward')
clusters = model.fit_predict(data)

print("Agglomerative Clustering:")
for i in range(len(movies)):
    print(movies[i], "-> Cluster", clusters[i] + 1)

# Dendrogram
linked = linkage(data, method='ward')

plt.figure(figsize=(8, 5))
dendrogram(linked, labels=movies)
plt.title("Agglomerative Hierarchical Clustering")
plt.xlabel("Movies")
plt.ylabel("Distance")
plt.show()


# -------------------------------
# Divisive Clustering
# -------------------------------

# Simple divisive approach:
# Start with all movies and split them into 2 groups
divisive = AgglomerativeClustering(
    n_clusters=2,
    linkage='ward'
)

div_clusters = divisive.fit_predict(data)

print("\nDivisive Clustering:")
for i in range(len(movies)):
    print(movies[i], "-> Group", div_clusters[i] + 1)
```
