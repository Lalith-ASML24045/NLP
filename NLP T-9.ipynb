```python
from collections import Counter

# Training corpus
words = ["I", "love", "NLP", "I", "love", "Python"]

# Count word frequencies
count = Counter(words)

vocab = len(count)
total = len(words)

# Laplace smoothing
print("Word Probabilities:")

for word in count:
    probability = (count[word] + 1) / (total + vocab)
    print(word, ":", probability)

# Probability of an unseen word
word = "AI"
probability = (0 + 1) / (total + vocab)

print("Unseen word probability:", probability)
```
